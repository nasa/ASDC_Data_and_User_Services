# Subsetting TEMPO Level 1 Mission Data

## Overview

Welcome to this tutorial on subsetting **TEMPO (Tropospheric Emissions: Monitoring of Pollution) Radiances (RAD) and Radiances Twilight (RADT) Level 1** data. Since TEMPO Level 1 products are more than 20 times larger than Level 2 products, subsetting the data by variable, temporal, and/or spatial parameters reduces transfer time and helps focus on the region or variables of interest.

### Prerequisites

**Harmony-py:**

Make sure the following packages are installed in your environment:
```bash
pip install harmony-py 
```
`harmony-py` is a package that allows for easy interaction with the Harmony API directly in Python.


**Authentication:**

The `harmony.Client` class will attempt to use credentials from a local `.netrc` file, located in the home directory of the filesystem where this notebook is running. Ensure your .netrc file contains Earthdata credentials for your Earthdata Login:

```
machine urs.earthdata.nasa.gov
    login <edl_username>
    password <edl_password>
```

````python
import os

from pathlib import Path

from harmony import BBox, Client, Collection, Request, Environment

# Create output directory
output_dir = Path("./TEMPO-output")
os.makedirs(output_dir, exist_ok=True)

op_dirs = []
````

## Initialize the Harmony client
Initialize the Harmony `Client` to interact with the NASA Earthdata production environment.

````python
harmony_client = Client(env=Environment.PROD)
````

## Define your TEMPO data request
Set up a Harmony `Request` specifying the TEMPO RAD Level-1 (L1) or TEMPO RADT L1 collection and the desired subsetting parameters. This example requests a set of variables over a small region of the Pacific Ocean off the southern coast of Alaska. Adjust the bounding box coordinates or variable list to your own area of interest.

````python
collection = Collection(id="C3868234842-LARC_CLOUD")  # TEMPO_RAD_L1
request = Request(
    collection=collection,
    spatial=BBox(-147.5466, 55.8881, -145.5466, 57.8881),  # Restrict data to this bounding box
    max_results=1,  # Small for demo purposes; increase as needed
    variables=[
        "band_290_490_nm/radiance",
        "band_290_490_nm/radiance_error",
        "band_290_490_nm/radiance_status_flag",
    ],
)

# Validate the request before submitting
print(f"Request is valid: {request.is_valid()}")
````

````python
# Submit the request and wait for processing to complete
job_id = harmony_client.submit(request)
````

## Download the output file(s)
Download the output file to the output directory `TEMPO-output`.

````python
downloaded_files = [
    f.result() for f in harmony_client.download_all(job_id, directory=output_dir, overwrite=True)
]
````
